# Deleting Redundant Columns

```python
data.columns
```

```python
columns_to_drop = ['id',  'reviews per month',
       'review rate number', 'calculated host listings count',
       'availability 365', 'house_rules', 'license']

data.drop(columns=columns_to_drop, inplace=True)
print(data.shape)
```

# Renaming Columns

```python
data.columns
```

```python
new_names = []
for col in data.columns:
    new_names.append(col.lower().capitalize())

new_names
```

```python
data.columns = new_names
data.columns
```

# Dropping Duplicates

```python
data.duplicated().value_counts()
```

```python
data.shape
```

```python
data.drop_duplicates(inplace = True)
data.shape
```

# Removing NaN values

```python
data.isna().sum()
```

```python
# Dropping Last Review Column
data.drop(columns = ["Last review"],inplace = True)
```

```python
data.isna().sum()
```

```python
data.dropna(inplace = True)
data.isna().sum()
```

```python
data.shape
```

# Clean Individual Columns

```python
data.sample(10)
```

```python
## Host_identity_verified column values will be converted to uppercase 
data["Host_identity_verified"] = data["Host_identity_verified"].str.upper()
data.sample(10) 
```

```python
## Converting Boolean Columns into Binary Columns
data["Instant_bookable"] = data["Instant_bookable"].map({True:1,False:0})
data.sample(10)
```

```python
# Removing Dollar Symbols
data["Service fee"] = data["Service fee"].str.replace("$","")
data.sample(10)
```

```python
data["Price"] = data["Price"].str.replace("$","")
data.sample(10)
```

# Converting columns Construction Year and Minimum Nights from float to integer

```python
data["Construction year"] = data["Construction year"].astype(np.int64)
data["Minimum nights"] = data["Minimum nights"].astype(np.int64)

data.sample(10)
```

```python
data["Number of reviews"] = data["Number of reviews"].astype(np.int64)
data.sample(10)
```
